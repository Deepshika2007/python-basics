```python
#Factorial of a number
def fac(n):
 if n==0 or n==1:
  return 1
 else:
  return n*fac(n-1)
n=int(input("enter the number"))
print (fac(n))
```

```python
#sum of n number
def natural(n):
   if (n==0):
    return 0
   if (n<0):
    return "please enter natural number"
   return n+natural(n-1)

n=int(input("enter the number"))
print (natural(n))
```

```python
# finfing second largest in array
arr=[2,4,20,22,14]
s_lar=0
lar=arr[0]
for x in arr:
  if (x>lar):
    s_lar=lar
    lar=x
  elif (x>s_lar):
    s_lar=x
print ("the second largest number is",s_lar)
```

```python
#to find largest and smallest in array
arr=[2,4,22,14,10]
largest=arr[0]
smallest=arr[0]
for x in arr:
  if x>largest:
    largest=x
  if x<smallest:
    smallest=x
print ("the largest is",largest)
print ("the smallest is",smallest)
```
